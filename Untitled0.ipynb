```python
!pip -q install transformers accelerate gradio

import torch, gradio as gr
from transformers import AutoTokenizer, AutoModelForCausalLM

MODEL = "Qwen/Qwen2.5-3B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForCausalLM.from_pretrained(
    MODEL, torch_dtype=torch.float16, device_map="auto"
)

SYSTEM = """You are a discrete mathematics tutor. Solve the question step by step:
1. State what is being asked and the topic (logic, sets, counting, graphs, number theory, etc.).
2. Show each step with the formula or rule used.
3. Double-check any arithmetic.
4. End with a line starting 'Final answer:'."""

def solve(question):
    if not question.strip():
        return "Please type a question."
    messages = [
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": question},
    ]
    prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    with torch.no_grad():
        out = model.generate(
            **inputs, max_new_tokens=700, do_sample=False
        )
    return tokenizer.decode(
        out[0][inputs["input_ids"].shape[1]:], skip_special_tokens=True
    )

gr.Interface(
    fn=solve,
    inputs=gr.Textbox(lines=3, label="Discrete math question"),
    outputs=gr.Markdown(label="Answer"),
    title="Discrete Math Helper",
    examples=[
        "How many ways can 5 people sit around a circular table?",
        "Find the number of edges in K_8.",
        "Is (p → q) ∧ (q → r) → (p → r) a tautology?",
        "How many 3-letter strings can be made from A-E with no repeats?",
    ],
).launch(share=True)
```
